```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

X1 = [[1,1,1]]
Y1 = [[1,2,2]]

print("Similarity score:", cosine_similarity(X1, Y1)[0][0])
```

# 1- Brute Force Numpy Implementation

```python
X = [1,1,1]
Y = [1,2,2]
```

### 1.1- dot product implementation

```python
def my_dot_product(X, Y):
    scalar = 0
    for x,y in zip(X,Y):
        scalar += x*y
        
    return scalar
```

```python
my_dot_product(X, Y)
```

### 1.2- vector norm implementation

```python
import math

def my_norm(X):
    norm = 0
    
    for x in X:
        norm += np.square(x)
        
    norm = math.sqrt(norm)
    return float(norm)
```

```python
my_norm(Y)
```

```python
def my_cosine_similarity(X, Y):
    norm_X, norm_Y = my_norm(X), my_norm(Y)
    
    if norm_X == 0 or norm_Y == 0:
        return 0.0
    
    cos_sim = my_dot_product(X, Y) / (norm_X * norm_Y)
    return float(cos_sim)

my_cosine_similarity(X, Y)

print("Similarity score:", my_cosine_similarity(X, Y))
```

```python
sk_one = cosine_similarity(X1, Y1)[0][0]
my_one = my_cosine_similarity(X, Y)
sk_one == my_one
```

# Scaled 1-to-N (Query vs. Matrix of Documents)

the similarity of:
- 1 query vector q of shape (D,) sent by the user
- against a database of N documents embeddings docs of shape (N, D)

</div>

```python
def query_to_docs_similarity(q: np.ndarray, docs: np.ndarray) -> np.ndarray:
    dots = docs @ q # (N,D) * (D,1) -> (N,1)
    Epsilon = 1e-12 # to prevent division by zero
    
    norm_docs = np.linalg.norm(docs, axis=1) #axis_1 for rows
    norm_q = np.linalg.norm(q)
    
    similarities = dots / (norm_docs * norm_q + Epsilon) # array [N] N score for N documents
    return similarities
```

# Top-K Retrieval (Vector Search)

```python
def top_k_retrieval(q: np.ndarray, docs: np.ndarray, k: int) -> np.ndarray:
    similarities = query_to_docs_similarity(q, docs)
    
    # np.argsort sorts ascending; [-k:] takes top k; [::-1] reverses to descending
    top_k_indices = np.argsort(similarities)[-k:][::-1]
    
    return top_k_indices
```

```python
docs_test = np.array([
    [1.0, 1.0, 1.0],
    [1.0, 2.0, 2.0],  # Doc 1 (identical to query!)
    [0.0, 5.0, 0.0],
])

q_test = np.array([1.0, 2.0, 2.0])
```

```python
print("All Similarities:", query_to_docs_similarity(q_test, docs_test))
retrieved = top_k_retrieval(q_test, docs_test, k=2)
print(f"Top 2 Matches are Doc {retrieved[0]} and Doc {retrieved[1]}", )
```
